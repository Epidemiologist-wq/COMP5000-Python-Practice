# Python Tuples

A **tuple** is used to store multiple values, similar to a list.

## Creating a Tuple

Tuples use **round brackets `( )`**.

```python
t = ("Jane", "Roger")
print(t)
```

Output:

```text
('Jane', 'Roger')
```

## Accessing Tuple Values

Like lists, tuples use **indexing**.

Python starts indexing at **0**.

```python
print(t[0])
print(t[1])
```

Output:

```text
Jane
Roger
```

## Important Difference: Tuple Cannot Be Changed

A tuple is **immutable**, which means its values cannot be changed after it is created.

For example:

```python
t[1] = "Peter"
```

This produces an error:

```text
TypeError: 'tuple' object does not support item assignment
```

This is because tuples cannot be changed.

## List vs Tuple

| | List | Tuple |
|---|---|---|
| Example | `[23, 14, 6, 8]` | `("Jane", "Roger")` |
| Brackets | `[ ]` | `( )` |
| Access values | ✅ Yes | ✅ Yes |
| Change values | ✅ Yes | ❌ No |

### When do we use a tuple?

Use a **tuple when you have a group of values that should not be changed**.

Example:

```python
date = (15, 6, 1995)
```

### Easy rule to remember

**List = changeable**

**Tuple = fixed / cannot be changed**

```python
# List
ages = [25, 31, 63]
ages[0] = 26       # ✅ allowed

# Tuple
date = (15, 6, 1995)
date[0] = 16       # ❌ not allowed
```

# Python `for` Loops

A **loop** is used when we want Python to repeat an action for multiple items.

## Basic pattern

```python
for item in collection:
    action
```

Think:

**FOR → each ITEM → IN → COLLECTION → do ACTION**

## Example

```python
v = [23, 14, 6, 8]

for number in v:
    print(number)
```

This prints every number in the list.

## Example with multiplication

```python
for number in v:
    print(5 * number)
```

The `5` is just a number chosen for the example. It is **not special to loops**.

The loop repeats the multiplication for every item:

```text
23 × 5 = 115
14 × 5 = 70
6 × 5 = 30
8 × 5 = 40
```

## Why use loops?

Loops save us from writing the same code repeatedly.

Instead of:

```python
print(v[0])
print(v[1])
print(v[2])
print(v[3])
```

we can use:

```python
for number in v:
    print(number)
```

### Key idea

> A loop is about **repetition**, not multiplication.

The action inside the loop can be printing, adding, multiplying, checking a condition, or another operation.

## 11. Mean / Average

The mean is:

**total ÷ number of values**

Using a loop:

```python
numbers = [10, 20, 30, 40]

total = 0

for number in numbers:
    total = total + number

mean = total / len(numbers)

print(mean)
```

NumPy can also calculate the mean:

```python
import numpy as np

mean = np.mean(numbers)
print(mean)
```

## 12. Dictionaries

A dictionary stores **key-value pairs**.

```python
marks = {
    "craig": 65,
    "vincent": 75
}
```

Get a value using its key:

```python
print(marks["craig"])
```

Add a new value:

```python
marks["muhammad"] = 80
```

Get all keys:

```python
marks.keys()
```

Get all values:

```python
marks.values()
```

## 13. While Loops

A `while` loop repeats while a condition is true.

Basic pattern:

```python
while condition:
    do_something
```

Example:

```python
number = int(input("Enter a number: "))

while number >= 0:
    number = int(input("Enter a number: "))
```

The loop stops when the user enters a negative number.
